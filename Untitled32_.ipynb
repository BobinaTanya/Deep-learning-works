```python
import gym
import numpy as np
import matplotlib.pyplot as plt

# Инициализация среды MountainCar
env = gym.make("MountainCar-v0")

# Параметры Q-Learning
ALPHA = 0.1        # Скорость обучения
GAMMA = 0.99       # Коэффициент дисконтирования
EPSILON = 0.1      # Эпсилон для epsilon-greedy стратегии
EPISODES = 1000    # Количество эпизодов
MAX_STEPS = 200    # Максимальное количество шагов в эпизоде

# Размеры дискретизации состояний
state_space_bins = [24, 24]

# Инициализация Q-таблицы
q_table = np.zeros(state_space_bins + [env.action_space.n])

# Функция для дискретизации состояний
def discretize_state(state):
    state_bins = [
        np.linspace(-1.2, 0.6, state_space_bins[0] + 1),  # Для позиции
        np.linspace(-0.07, 0.07, state_space_bins[1] + 1)  # Для скорости
    ]
    discrete_state = [
        np.digitize(state[0], state_bins[0]) - 1,  # Позиция
        np.digitize(state[1], state_bins[1]) - 1   # Скорость
    ]
    return tuple(discrete_state)

# Функция для выбора действия по стратегии epsilon-greedy
def choose_action(state):
    if np.random.rand() < EPSILON:  # Исследование (с вероятностью epsilon)
        return np.random.choice(env.action_space.n)  # Случайное действие
    else:  # Эксплуатация (выбор наилучшего действия)
        return np.argmax(q_table[state])  # Действие с максимальным значением Q

# Основной цикл обучения
reward_list = []  # Список для сохранения награды каждого эпизода

for episode in range(EPISODES):
    state = env.reset()  # Сброс среды
    state = discretize_state(state)
    total_reward = 0

    for step in range(MAX_STEPS):
        action = choose_action(state)  # Выбор действия
        next_state, reward, done, _ = env.step(action)  # Выполнение действия

        # Дискретизация следующего состояния
        next_state = discretize_state(next_state)

        # Обновление Q-таблицы
        q_table[state][action] += ALPHA * (reward + GAMMA * np.max(q_table[next_state]) - q_table[state][action])

        state = next_state  # Переход в следующее состояние
        total_reward += reward  # Подсчет общей награды

        if done:
            break

    reward_list.append(total_reward)  # Добавление награды эпизода в список

# Построение графика награды за эпизоды
plt.plot(range(EPISODES), reward_list)
plt.xlabel('Эпизоды')
plt.ylabel('Награда')
plt.title('Награды за эпизоды (Q-Learning для MountainCar)')
plt.show()
```
